```python
import requests
```

```python
response = requests.post(
    "http://localhost:11434/api/generate",
    json={
        "model": "llama3.2:3b",
        "prompt": "Explain what data science is in simple terms.",
        "stream": False
    }
)

result = response.json()
result
```

```python
print("Modèle :", result["model"])
print("Réponse :", result["response"])
print("Tokens du prompt :", result["prompt_eval_count"])
print("Tokens générés :", result["eval_count"])
```

```python
def generate(prompt):
    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": "llama3.2:3b",
            "prompt": prompt,
            "stream": False
        }
    )

    return response.json()
```
